### 1.4. String Methods

```python
# Các methods hữu ích cho strings
name = input("What's your name? ")

# Xóa khoảng trắng thừa và viết hoa chữ cái đầu
name = name.strip().title()

print(f"hello, {name}")
```

```python
# Demo các string methods khác
text = "  hello world  "

print(f"Original: '{text}'")
print(f"strip(): '{text.strip()}'")
print(f"upper(): '{text.upper()}'")
print(f"lower(): '{text.lower()}'")
print(f"title(): '{text.title()}'")
print(f"capitalize(): '{text.capitalize()}'")
```

## Phần 2: Integers và Floats

### 2.1. Integers (int)

```python
# Calculator đơn giản
x = int(input("What's x? "))
y = int(input("What's y? "))

z = x + y

print(f"Result: {z}")
```

**Phân tích:**
- `int()` chuyển đổi (cast) string sang integer
- Mặc định `input()` trả về string, phải cast sang int để tính toán
- Các phép toán: `+`, `-`, `*`, `/`, `//` (chia lấy phần nguyên), `%` (chia lấy dư), `**` (lũy thừa)

```python
# Demo các phép toán với integers
a = 10
b = 3

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")  # Chia lấy phần nguyên
print(f"{a} % {b} = {a % b}")    # Chia lấy dư (modulo)
print(f"{a} ** {b} = {a ** b}")  # Lũy thừa
```

### 2.2. Floats

```python
# Làm việc với số thập phân
x = float(input("What's x? "))
y = float(input("What's y? "))

z = x / y

# Format với 2 chữ số thập phân
print(f"Result: {z:.2f}")
```

**Phân tích:**
- `float()` chuyển đổi sang số thập phân
- `{z:.2f}` format với 2 chữ số thập phân
- `round(z, 2)` là cách khác để làm tròn

```python
# Demo formatting numbers
number = 1234.56789

print(f"Original: {number}")
print(f"2 decimals: {number:.2f}")
print(f"With comma: {number:,.2f}")
print(f"Rounded: {round(number, 2)}")
```

### 4.4. AND Operator

```python
# Hệ thống chấm điểm
score = int(input("Score: "))

if score >= 90 and score <= 100:
    print("Grade: A")
elif score >= 80 and score < 90:
    print("Grade: B")
elif score >= 70 and score < 80:
    print("Grade: C")
elif score >= 60 and score < 70:
    print("Grade: D")
else:
    print("Grade: F")
```

**Phân tích:**
- `and`: True chỉ khi TẤT CẢ điều kiện đúng
- Python hỗ trợ cú pháp đặc biệt: `90 <= score <= 100`

```python
# Cách viết Pythonic hơn
score = int(input("Score: "))

if 90 <= score <= 100:
    print("Grade: A")
elif 80 <= score < 90:
    print("Grade: B")
elif 70 <= score < 80:
    print("Grade: C")
elif 60 <= score < 70:
    print("Grade: D")
else:
    print("Grade: F")
```

### 4.5. Modulo Operator

```python
# Kiểm tra số chẵn/lẻ
x = int(input("What's x? "))

if x % 2 == 0:
    print("Even")
else:
    print("Odd")
```

**Phân tích:**
- `%` (modulo) trả về phần dư của phép chia
- `x % 2 == 0`: số chẵn (chia hết cho 2)
- `x % 2 != 0`: số lẻ (không chia hết cho 2)

```python
# Demo modulo
print(f"10 % 3 = {10 % 3}")  # 1 (10 = 3*3 + 1)
print(f"15 % 4 = {15 % 4}")  # 3 (15 = 4*3 + 3)
print(f"20 % 5 = {20 % 5}")  # 0 (20 = 5*4 + 0)
```

### 4.6. Boolean Functions

```python
# Tạo function trả về boolean
def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False

def main():
    x = int(input("What's x? "))
    if is_even(x):
        print("Even")
    else:
        print("Odd")

main()
```

**Phân tích - Cách viết Pythonic hơn:**

```python
# Pythonic version 1: Ternary operator
def is_even(n):
    return True if n % 2 == 0 else False

# Pythonic version 2: Trực tiếp (TỐT NHẤT)
def is_even(n):
    return n % 2 == 0  # Tự động trả về True/False

print(is_even(4))  # True
print(is_even(7))  # False
```

### 4.7. Match Statements

```python
# Match statement (Python 3.10+)
name = input("What's your name? ")

match name:
    case "Harry":
        print("Gryffindor")
    case "Hermione":
        print("Gryffindor")
    case "Ron":
        print("Gryffindor")
    case "Draco":
        print("Slytherin")
    case _:
        print("Who?")
```

**Phân tích:**
- `match` tương tự `switch` trong các ngôn ngữ khác
- `case _`: default case (bắt tất cả)
- Có thể dùng `|` để combine nhiều cases

```python
# Match với multiple values
name = input("What's your name? ")

match name:
    case "Harry" | "Hermione" | "Ron":
        print("Gryffindor")
    case "Draco":
        print("Slytherin")
    case _:
        print("Who?")
```

### 6.3. Error Handling Preview

```python
# Xử lý lỗi cơ bản (sẽ học chi tiết ở bài sau)
def safe_divide(x, y):
    """Chia an toàn, tránh chia cho 0."""
    if y == 0:
        return "Cannot divide by zero"
    return x / y

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # Cannot divide by zero
```
